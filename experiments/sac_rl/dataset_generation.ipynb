```python
# Make the repo-root modules (simulator.py, utils.py) importable from this folder
import sys, os
sys.path.insert(0, os.path.abspath('../..'))
```

```python
import pickle
from utils import make_training_dataset
```

```python
Lx, Ly = 2.0, 1.0
nx, ny = 100, 50
dt = 0.004

num_samples = 1000
num_sources_sensors = {(1, 3): 0.2, (1, 4): 0.2, (2, 5): 0.6}
noise_std = {0.0: 0.5, 0.1: 0.5}
nt = {800: 0.6, 1000: 0.2, 1200: 0.2}
kappa = {0.05: 0.5, 0.1: 0.5}
bc = "random"
T0 = None
q_range = (0.5, 2)
```

```python
# Generate dataset using the helper function
dataset = make_training_dataset(
    Lx=Lx, Ly=Ly, nx=nx, ny=ny, num_samples=num_samples,
    dt=dt, num_sources_sensors=num_sources_sensors, q_range=q_range,
    noise_std=noise_std, nt=nt, kappa=kappa, bc=bc, T0=T0, seed=42
)

# Save as pickle (not numpy!)
os.makedirs("data", exist_ok=True)
with open("data/training_dataset.pkl", "wb") as f:
    pickle.dump(dataset["samples"], f)
```
